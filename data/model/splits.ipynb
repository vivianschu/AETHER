```python
import pandas as pd
```

```python
# split mhc_set.csv into train, test, valid sets
df = pd.read_csv("mhc_set.csv")
train = df.sample(frac=0.8, random_state=42)
test_valid = df.drop(train.index)
test = test_valid.sample(frac=0.5, random_state=42)
valid = test_valid.drop(test.index)

# print shapes
print(train.shape)
print(test.shape)
print(valid.shape)

folder = "splits/mhc-80-10-10"
train.to_csv(f"{folder}/train.csv", index=False)
test.to_csv(f"{folder}/test.csv", index=False)
valid.to_csv(f"{folder}/valid.csv", index=False)
```
